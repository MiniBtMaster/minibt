# btplot 一图看懂：K 线 / 指标 / 信号

`btplot()` 是 minibt 的"一行出图"入口：把 K 线、指标线、买卖信号画在同一张图上，支持三种引擎（原生窗口 / 浏览器 / Notebook）。本篇用最少的代码把**四种常见画法**各跑一次。

**本次演示**

- `KLine.btplot()` 画裸 K 线（不带指标）
- 链式指标 + `btplot()` 一次显示多条线
- `signal_only` / `overlap` 控制信号与副图
- 自定义 `BtIndicator`（class MA）也能直接 `btplot()`

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[1.17minibt_interactive_financial_charting_with_btplot.md](../minibt_docs/docs/minibt_basic/1.17minibt_interactive_financial_charting_with_btplot.md)


## 1. 导入与内置数据


```python
from minibt import *
# 导入本地数据
kline=LocalDatas.test.kline(height=400)
```

## 2. 裸 K 线：btplot()


```python
# 绘制K线图
kline.btplot()
```

## 3. 链式赋值：一次画多条指标


```python
# 链式赋值并显示所有指标
gc=kline.tradingview.G_Channels().avg.ema().ebsw().btplot(include="all",black_style=True)
```

## 4. 交易信号：long_signal / short_signal


```python
# 交易信号显示
uba=kline.tradingview.UT_Bot_Alerts()
uba.btplot()
```

## 5. 副图指标（pmax3）


```python
# pmax3指标显示
kline.btind.pmax3().btplot()
```

## 6. 自定义 BtIndicator：class MA


```python
class MA(BtIndicator):
    params=dict(length1=3,length2=5)
    isplot = dict(long_signal=False, short_signal=False)
    overlap = dict(ma3=True, ma5=True)

    def next(self):
        ma3 = self.close.sma(self.params.length1)
        ma5 = self.close.sma(self.params.length2)
        long_signal = ma3.cross_up(ma5)
        short_signal = ma3.cross_down(ma5)
        return ma3, ma5, long_signal, short_signal
    
ma=MA(kline)
ma.btplot()
```

## 小结

- `btplot()` 既能在挂上指标后调用，也能直接 `kline.btplot()`；
- 信号只要写进 `long_signal` / `short_signal` 就会自动画箭头；
- 想换引擎只改 `gui=`：`"pylightcharts"` / `"bokeh"` / `"jupyter"`。
