# PandasTa 全量：174 个指标逐个跑

`kline.pta` 是默认指标库（pandas_ta 口径）。本篇用**默认参数**把全部指标跑一遍，顺便看哪些需要额外参数。

**本次演示**

- `get_pandas_explicit_methods()` 取出显式方法清单
- 默认参数下能算 / 不能算的分开记
- `other` 里是需要额外序列或特殊参数的算子
- 最后用 `cross()` 组合两条线

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[pandasta.md](../minibt_docs/docs/minibt_api_reference/pandasta.md)


## 1. 导入 minibt


```python
from minibt import *
options.set_conversion_mode="free"
```

## 2. 内置数据（前 1000 根）


```python
kline=LocalDatas.test.kline.iloc[:1000]
kline.head()
```

## 3. 清单工具：get_pandas_explicit_methods()


```python
from minibt.utils import get_pandas_explicit_methods
pandasta_method=list(get_pandas_explicit_methods(PandasTa))
for i in range(0, len(pandasta_method), 10):
    print(pandasta_method[i:i+10])
```

## 4. 默认参数逐个跑一遍


```python
default_method=[]
other=[]
for method in pandasta_method:
    try:
        ind=getattr(kline,method)()
        print(f"指标{method},lines:{ind.lines},最后的数据:{ind.values[-1]}")
        default_method.append(method)
    except:
        other.append(method)
print("默认参数可运行指标:")
for i in range(0, len(default_method), 10):
    print(default_method[i:i+10])
```

## 5. 需要额外参数的算子


```python
other
```

## 6. 组合用法示例


```python
print(kline.close.cross(kline.close.ema(10)))
print(kline.close.above(kline.close.ema(10)))
print(kline.cdl_pattern("2crows"))
print(kline.close.cross_up(kline.close.ema(10)))
print(kline.close.below(kline.close.ema(10)))
print(kline.close.cross_down(kline.close.ema(10)))

```

## 小结

- pandas_ta 的口径是 `length`；和 `talib` 的 `timeperiod` 不同；
- 默认参数失败通常是因为需要第二条序列（`series=`）。
