# TqFunc 全量：28 个天勤函数

`kline.tqfunc` 是天勤的**函数式**接口（小写 + `length`），返回 `IndSeries`，适合链式组合。

**本次演示**

- 函数式与门面式的区别
- `crossup` / `crossdown` 组合用法
- 和指标库同名函数的对照

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[tqfunc.md](../minibt_docs/docs/minibt_api_reference/tqfunc.md)


## 1. 导入 minibt


```python
from minibt import *
options.set_conversion_mode="free"
```

## 2. 内置数据（前 1000 根）


```python
kline=LocalDatas.test.kline.iloc[:1000]
kline.head()
```

## 3. 清单工具：get_pandas_explicit_methods()


```python
from minibt.utils import get_pandas_explicit_methods
pandasta_method=list(get_pandas_explicit_methods(TqFunc))
for i in range(0, len(pandasta_method), 10):
    print(pandasta_method[i:i+10])
```

## 4. 默认参数逐个跑一遍


```python
default_method=[]
other=[]
for method in pandasta_method:
    try:
        ind=getattr(kline.tqfunc,method)()
        print(f"指标{method},lines:{ind.lines},最后的数据:{ind.values[-1]}")
        default_method.append(method)
    except:
        other.append(method)
print("默认参数可运行指标:")
for i in range(0, len(default_method), 10):
    print(default_method[i:i+10])
```

## 5. 需要额外参数的算子


```python
other
```

## 6. 组合用法示例


```python
print(kline.close.sma(5).tqfunc.crossup(kline.close.sma(10)))
print(kline.close.sma(5).tqfunc.crossdown(kline.close.sma(10)))
print(kline.close.tqfunc.min(kline.open))
print(kline.close.tqfunc.max(kline.open))
print((kline.close.sma(5)>kline.close.sma(10)).tqfunc.cum_counts())
print((kline.close.sma(5)>kline.close.sma(10)).tqfunc.every())
```

## 小结

- 函数式接口更贴近 TqSdk 文档，策略里可读性好；
- 结果类型与门面式完全一致（都是 IndSeries）。
