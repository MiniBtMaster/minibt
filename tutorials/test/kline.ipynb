# KLine 全功能速查：属性 / 重采样 / 回放 / 运行

`KLine` 是整套框架的数据根节点：行情、账户、订单、回放、指标都挂在它上面。本篇把**常用能力**逐条跑一遍，当作速查表。

**本次演示**

- `broker` / `TargetPosTask` 等交易接口
- `resample()` / `replay()`
- `run()` 跑指标
- 蜡烛图样式开关（`Heikin_Ashi_Candles` 等）

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[9.3_kline.md](../minibt_docs/docs/minibt_data/9.3_kline.md)


## 1. 导入与内置数据


```python
from minibt import *
kline=LocalDatas.test.kline
kline.head()
```

## 2. 账户 / 下单接口


```python
(kline.broker
,kline.orders
,kline.follow
,kline.current_datetime
,kline.current_time
,kline.current_open
,kline.current_high
,kline.current_low
,kline.current_close
,kline.symbol_info
,kline.symbol
,kline.cycle
,kline.price_tick
,kline.volume_multiple
,kline.quote
,kline.commission
,kline.tick_commission
,kline.percent_commission
,kline.fixed_commission
,kline.slip_point
,kline.stop
,kline.stop_lines
,kline.stop_price
,kline.target_price
,kline.margin
,kline.step_margin
,kline.position_margin
,kline.margin_rate
,kline.position
,kline.float_profit
,kline.float_tick
,kline.open_price
,kline.cost_price
,kline.datetime
,kline.open
,kline.high
,kline.low
,kline.close
,kline.volume
,kline.bear_color
,kline.bull_color
,kline.datetime_index)




```

## 3. 目标仓位


```python
(kline.TargetPosTask(1)
,kline.buy()
,kline.sell()
,kline.set_target_size()
)
```

## 4. 重采样 resample()


```python
resample_kline=kline.resample(60*15,height=400)
resample_kline.sma(30).btplot()
```

## 5. 回放 replay()


```python
replay_kline=kline.replay(60*15,height=400)
replay_kline.close.sma(overlap=True).btplot("last")
```

## 6. run() 跑指标


```python
kline.run(["ma1", Multiply(PandasTa.sma, dict(length=20))],
        ["ma2", Multiply(PandasTa.sma, dict(length=30))],
        ["ma3", Multiply(PandasTa.sma, dict(length=60))],
        ["ma4", Multiply(PandasTa.sma, dict(length=120))],
        ["ebsw", Multiply(PandasTa.ebsw)])
```

## 7. Heikin-Ashi 蜡烛


```python
kline.Heikin_Ashi_Candles=True
kline.btplot()
```

## 8. 线性回归蜡烛


```python
kline.Linear_Regression_Candles=True
kline.btplot()
```

## 小结

- KLine 既能当"数据容器"，也能当"交易入口"；
- `resample()` / `replay()` 都返回**新的 KLine**，原对象不变。
