# `kline.run()`：把指标 / 策略 / 回测一次跑完

`run()` 是 K 线对象的"跑一批东西"入口：可以传指标名、`Modify(...)` 表达式、策略子类，顺序执行并出图。

**本次演示**

- `run(["ma1", Multiply(PandasTa.sma, dict(length=20))])` 的写法
- 一次挂多个指标（列表）
- `run()` 之后 `btplot()` 看结果

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[1.11minibt_internal_data_btdata_guide.md](../minibt_docs/docs/minibt_basic/1.11minibt_internal_data_btdata_guide.md)


## 1. 导入与内置数据


```python
from minibt import *
kline=LocalDatas.test.kline.iloc[:1000]
kline.head()
```

## 2. run() + btplot()


```python
kline.run(["ma1", Multiply(PandasTa.sma, dict(length=20))],
        ["ma2", Multiply(PandasTa.sma, dict(length=30))],
        ["ma3", Multiply(PandasTa.sma, dict(length=60))],
        ["ma4", Multiply(PandasTa.sma, dict(length=120))],
        ["ebsw", Multiply(PandasTa.ebsw)])
# 如果没图表的话请重新运行单元格
```

## 小结

- `run()` 只负责"算"，画图仍交给 `btplot()`；
- `Multiply(...)` 用来把库方法包成"可复用的指标工厂"。
