# IndFrame 上的 pandas 全家桶（78 个方法）

`IndFrame` 是 DataFrame 的子类，pandas 的表格 API 几乎都能用。本篇 78 个单元格按"构造 / 增删列 / 清洗 / 排序 / 统计 / 转换"分组过一遍，当速查表用。

**本次演示**

- 构造：空 / dict / 形状 / 从 DataFrame
- 增删改：`assign` / `insert` / `drop` / `pop` / `eval`
- 清洗：`fillna` / `dropna` / `replace` / `clip` / `round`
- 统计：`mean` / `corr` / `rank` / `rolling` 等

## 小结

- `IndFrame` 与 pandas 的差别只有一处：**列名必须在 `lines` 里**，否则画图会漏线；
- 大表建议 `IndFrame(df)` 一次构造，别逐列赋值（慢）。
