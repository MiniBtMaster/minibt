## 1. 导入 minibt


```python
from minibt import *
options.set_conversion_mode="free"
```

## 2. 内置数据（前 1000 根）


```python
df=LocalDatas.test.kline.iloc[:1000]
df.head()
```

## 3. 清单工具：get_pandas_explicit_methods()


```python
kline=KLine(df)
kline.head()
```

## 4. 默认参数逐个跑一遍


```python
from minibt.utils import get_pandas_explicit_methods
pandasta_method=list(get_pandas_explicit_methods(TradingView))
for i in range(0, len(pandasta_method), 4):
    print(pandasta_method[i:i+4])
```

## 5. 需要额外参数的算子


```python
default_method=[]
other=[]
for method in pandasta_method:
    try:
        ind=getattr(kline.tradingview,method)()
        print(f"指标{method},lines:{ind.lines},最后的数据:{ind.values[-1]}")
        default_method.append(method)
    except:
        other.append(method)
print("默认参数可运行指标:")
for i in range(0, len(default_method), 10):
    print(default_method[i:i+10])
```
