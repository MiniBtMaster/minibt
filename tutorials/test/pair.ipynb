# Pair 指标全量：配对 / 协整 / 卡尔曼

配对交易要的统计量（z-score、协整、Hurst、卡尔曼…）都在 `kline.pairtrading` 里，一次跑完全部算子看结果。

**本次演示**

- 需要两条序列的算子怎么传（`kline1`）
- 哪些算子在 `NON_KLINE_INDICATORS`（不能直接画图）
- `vecm_based` / `kalman_filter` 的额外依赖

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[pair.md](../minibt_docs/docs/minibt_api_reference/pair.md)


## 1. 导入 minibt


```python
from minibt import *
options.set_conversion_mode="free"
```

## 2. 内置数据（前 1000 根）


```python
kline,kline1=LocalDatas.pp2609_60.kline.iloc[:1000],LocalDatas.l2609_60.kline.iloc[:1000]
kline.head(),kline1.head()
```

## 3. 清单工具：get_pandas_explicit_methods()


```python
from minibt.utils import get_pandas_explicit_methods
pandasta_method=list(get_pandas_explicit_methods(Pair))
for i in range(0, len(pandasta_method), 10):
    print(pandasta_method[i:i+10])
```

## 4. 默认参数逐个跑一遍


```python
default_method=['z_score', 'garch_volatility_adjusted', 'vecm_based', 'hurst_filter', 'rolling_quantile', 'kalman_filter', 'percentage_deviation', 'bollinger_bands']
spread=kline.close-kline1.close
for method in default_method:
    try:
        ind=getattr(spread.pairtrading,method)()
        print(f"指标{method},lines:{ind.lines},最后的数据:{ind.values[-1]}")
    except:
        ...
```

## 5. 需要额外参数的算子


```python
other=['vecm_based', 'kalman_filter']
for method in other:
    try:
        ind=getattr(kline.close.pairtrading,method)(kline1.close)
        print(f"指标{method},lines:{ind.lines},最后的数据:{ind.values[-1]}")
    except:
        ...
```
