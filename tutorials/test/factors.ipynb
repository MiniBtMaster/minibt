# Factors 全量：高阶因子工具

`kline.factors` 是"多因子建模"工具集：`adaptive_weight_trend` 等，输入可以是任意 `IndSeries` / `IndFrame`。

**本次演示**

- 方法清单与依赖（部分需要 sklearn / scipy）
- 在 `IndSeries` 上直接调用（`rsi.factors.xxx()`）
- `overlap=True` 让结果叠在主图

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[factors.md](../minibt_docs/docs/minibt_api_reference/factors.md)


## 1. 导入 minibt


```python
from minibt import *
options.set_conversion_mode="free"
```

## 2. 内置数据（前 1000 根）


```python
kline=LocalDatas.test.kline.iloc[:1000]
kline.head()
```

## 3. 清单工具：get_pandas_explicit_methods()


```python
from minibt.utils import get_pandas_explicit_methods
pandasta_method=list(get_pandas_explicit_methods(Factors))
for i in range(0, len(pandasta_method), 10):
    print(pandasta_method[i:i+10])
```

## 4. 默认参数逐个跑一遍


```python
rsi=kline.rsi()
zscore=kline.zscore()
macd=kline.macd()
adx=kline.adx()
pandasta_method.remove("adaptive_weight_trend")

for method in pandasta_method:
    ind=getattr(kline.close.factors,method)(rsi,zscore,macd,adx)
    print(f"指标{method},lines:{ind.lines},最后的数据:{ind.values[-1]}")


```

## 5. 需要额外参数的算子


```python
awt=kline.close(overlap=True).sma(10).factors.adaptive_weight_trend(overlap=True)
awt.get_indicator_kwargs(),awt.btplot()
```

## 小结

- Factors 返回 `IndFrame`（可能多列）；
- 自适应权重类因子需要足够长的历史（建议 ≥ 200 根）。
