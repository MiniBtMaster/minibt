# IndSeries 上的 pandas 全家桶（43 个方法）

`IndSeries` 继承了 pandas 的绝大部分方法：清洗、排序、变换、累积量……本篇 43 个单元格逐个走一遍，当速查表用。

**本次演示**

- 缺失值：`fillna` / `ffill` / `bfill` / `interpolate`
- 变换：`clip` / `where` / `mask` / `rank` / `pct_change`
- 归约排序：`sort_values` / `argsort` / `cummax`
- 转换：`reset_index` / `to_frame` / `__neg__`

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[1.12minibt_internal_data_series_guide.md](../minibt_docs/docs/minibt_basic/1.12minibt_internal_data_series_guide.md)


## 1. 导入


```python
from minibt import *
options.set_conversion_mode="free"
```

## 2. 取负


```python
series:IndSeries=IndSeries(range(10))
series,type(series)
```

## 3. take()


```python
series=-series
series1=series.abs()
series,series1,type(series)
```

## 4. filter()


```python
series.take(range(5))
```

## 5. 自定义函数管道


```python
series.filter(items=range(5))
```

## 6. astype 布尔


```python
def normalize_data(df):
    return (df - df.mean()) / df.std()
series.pipe(normalize_data)
```

## 7. copy()


```python
series.astype(np.bool_,True),series
```

## 8. 切片置 NaN


```python
series1=series.copy()
series2=series.copy(True)
series1,type(series1),series2,type(series2),series is series1,series is series2,series1 is series2
```

## 9. fillna


```python
series.iloc[:5]=np.full(5,np.nan)
series
```

## 10. 改单点


```python
series.fillna(0.)
```

## 11. ffill


```python
series.iloc[0]=100
series
```

## 12. bfill


```python
series.ffill()
```

## 13. 查看


```python
series.bfill()
```

## 14. replace


```python
series
```

## 15. interpolate


```python
series.replace(100,0)
```

## 16. clip


```python
series.interpolate()
```

## 17. 置 100


```python
series.clip(-50,50)
```

## 18. rank(pct)


```python
series.iloc[:5]=100
series
```

## 19. where


```python
series.rank(pct=True)
```

## 20. mask


```python
series.where(series>50.)
```

## 21. pct_change


```python
series.mask(series>50,1000)
```

## 22. reset_index


```python
series.pct_change()
```

## 23. to_frame


```python
series1=series.reset_index()#drop=False时转为IndFrame，但有整数索引
series1,type(series1),series1.columns,series1.pop(0)
```

## 24. duplicated


```python
series.to_frame(name='A')
```

## 25. round


```python
series.duplicated()
```

## 26. diff


```python
series.round()
```

## 27. copy(True)


```python
series.diff()
```

## 28. update


```python
series1=series.copy(True)
series1.iloc[-3:]=10
series1
```

## 29. sort_values


```python
series.update(series1)
series
```

## 30. sort_index


```python
series.sort_values()
```

## 31. argsort


```python
series.sort_index()
```

## 32. map


```python
series.argsort()
```

## 33. apply


```python
series.map(lambda x:x+1)
```

## 34. isin


```python
series.apply(lambda x:np.power(x,2))
```

## 35. between


```python
series.isin([100,])
```

## 36. case_when


```python
series.between(0,50)
```

## 37. isna


```python
series.case_when([
    (series<0,0),
    (series<10,10),
    (series>50,50)]
)
```

## 38. 置 NaN


```python
series.isna()
```

## 39. notna


```python
series.iloc[:3]=np.nan
series.isna()
```

## 40. notnull


```python
series.notna()
```

## 41. 置 1


```python
series.notnull()
```

## 42. cummax


```python
series.iloc[:3]=1
series
```

## 43. `print(…)`


```python
print(series.cummax())
print(series.cummin())
print(series.cumsum())
print(series.cumprod())
```
