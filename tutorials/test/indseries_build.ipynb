# IndSeries 构造与赋值：手工造一条线

从零构造 `IndSeries`：切片赋值、改元素、声明 `lines`，再挂到 K 线上画出来。

**本次演示**

- `IndSeries()` 空对象 → `series[:] = ...` 赋值
- `series.iloc[9] = 100` 改单点
- `series.kline` / `source_object` 看它挂在哪
- `lines=[...]` 决定图例名

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[9.5_indseries.md](../minibt_docs/docs/minibt_data/9.5_indseries.md)


## 1. 导入


```python
from minibt import *
```

## 2. 空 IndSeries


```python
series = IndSeries()
```

## 3. 切片赋值


```python
series[:]=range(10)
series,type(series)
```

## 4. 查 kline / source


```python
print(series.kline,series.source_object)
```

## 5. 画图


```python
series.btplot()
```

## 6. 改单点


```python
series.iloc[9]=100
series
```

## 7. 重新赋值


```python

series[:]=range(100)
series
```

## 8. 改图例名


```python
series.lines=["series"]
series.get_indicator_kwargs()
```

## 小结

- 手工构造的 IndSeries 没有 `kline`（除非显式挂上去），画图时要传 K 线；
- `lines` 长度必须为 1（单线指标）。
