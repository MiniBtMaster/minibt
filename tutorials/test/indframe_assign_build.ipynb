# IndFrame 构造与赋值：从零拼一个指标

不继承 `BtIndicator` 也能造指标：直接 `IndFrame(...)` + 赋值 + 声明 `lines` / `overlap` / 样式，然后挂到 K 线上。

**本次演示**

- `IndFrame()` / `IndFrame(dict(...))` 两种构造
- `df.close = ...` 与 `df["open"] = ...` 两种赋值
- `get_indicator_kwargs()` 看当前指标设置
- `overlap` / `lines` 决定主图还是副图

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[9.4_indframe.md](../minibt_docs/docs/minibt_data/9.4_indframe.md)


## 1. 导入


```python
from minibt import *
```

## 2. 空 IndFrame


```python
df=IndFrame()
df.head()
```

## 3. 属性赋值


```python
df.close=range(100)
print(df.close,type(df.close))
```

## 4. 下标赋值


```python
df["open"]=range(100)
print(df.open,type(df.open))
```

## 5. 从 KLine 派生


```python
kline = LocalDatas.test.kline
tick = kline.price_tick
ind = IndFrame()

# ===== 构建信号 =====
ma_fast = kline.close.sma(5)
ma_slow = kline.close.sma(20)
long_sig = ma_fast.cross_up(ma_slow)
short_sig = ma_fast.cross_down(ma_slow)
ind.ma_fast=ma_fast
ind.ma_slow=ma_slow
ind.long_signal = long_sig
ind.short_signal = short_sig

# ===== 构建附加参数 =====
# n = len(kline)
# atr_arr   = ind.close.atr(14).to_numpy()         # 数组①: 每 bar 的 ATR
# vol_arr   = ind.close.sma(20).to_numpy()          # 数组②: 均线基准
# min_tick  = 3.0                                    # 标量③: 最小保底 tick 数
# max_tick  = 50.0                                   # 标量④: 最大止损 tick 上限
# pct_limit = 0.03                                   # 标量⑤: 亏损百分比阈值
ind.tail(),ma_fast.kline
```

## 6. 查指标设置


```python
ind.get_indicator_kwargs()
```

## 7. source_object


```python
print(ind.source_object)
```

## 8. 改 overlap


```python
ind.ma_fast.overlap=True
ind.ma_slow.overlap=True
ind.btplot()
```

## 小结

- 手工构造适合"一次性计算"，长期复用建议写成 `BtIndicator`；
- `lines` 必须和数据列数对齐，否则画图会错位。
