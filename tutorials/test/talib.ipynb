# TaLib 全量：99 个指标逐个跑

`kline.talib` 是 TA-Lib 口径（**大写方法名** + `timeperiod`）。本篇跑完全部指标，并和 PandasTa 对照。

**本次演示**

- 大写方法名与 `timeperiod` 参数
- 哪些指标需要额外输入（`high` / `low` / `volume`）
- `MaType` 之类的枚举参数

不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[talib.md](../minibt_docs/docs/minibt_api_reference/talib.md)


## 1. 导入 minibt


```python
from minibt import *
options.set_conversion_mode="free"
```

## 2. 内置数据（前 1000 根）


```python
kline=LocalDatas.test.kline.iloc[:1000]
kline.head()
```

## 3. 清单工具：get_pandas_explicit_methods()


```python
from minibt.utils import get_pandas_explicit_methods
pandasta_method=list(get_pandas_explicit_methods(TaLib))
for i in range(0, len(pandasta_method), 10):
    print(pandasta_method[i:i+10])
```

## 4. 默认参数逐个跑一遍


```python
default_method=[]
other=[]
for method in pandasta_method:
    try:
        ind=getattr(kline,method)()
        print(f"指标{method},lines:{ind.lines},最后的数据:{ind.values[-1]}")
        default_method.append(method)
    except:
        other.append(method)
print("默认参数可运行指标:")
for i in range(0, len(default_method), 10):
    print(default_method[i:i+10])
```
