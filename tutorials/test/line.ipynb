# Line 速查：单条线的样式与信号

`Line` 是"图上的**一条线**"，能单独改线型、单独挂信号、单独控制是否跟随价格轴。

**本次演示**

- 从指标里取线：`pmax.long` / `pmax.thrend`
- `line_style` 改线型
- `signal_style` + `set_label()` 改箭头
- `overlap` / `isplot` / `follow` 三个开关

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[1.13minibt_internal_data_line_guide.md](../minibt_docs/docs/minibt_basic/1.13minibt_internal_data_line_guide.md)


## 1. 导入与内置数据


```python
from minibt import *
kline=LocalDatas.test.kline
kline.head()
```

## 2. 造 pmax 指标


```python
pmax=kline.close.btind.pmax()
long,short,thrend=pmax
long,type(long),long.source is pmax
```

## 3. 取一条线


```python
(pmax.long.line_style
,pmax.long.line_dash
,pmax.long.line_width
,pmax.long.line_color)
```

## 4. 改线型


```python
pmax.long.line_style=LineStyle()
pmax.long.line_dash=LineDash.dashdot
pmax.long.line_width=3.
pmax.long.line_color=Colors.red
```

## 5. 造信号（交叉）


```python
pmax.long_signal=(pmax.thrend== 1) & (pmax.thrend.shift()==-1)
pmax.short_signal=(pmax.thrend== -1) & (pmax.thrend.shift()==1)
pmax.long_signal.isplot=False
pmax.short_signal.isplot=False

```

## 6. 看信号样式


```python
(pmax.long_signal.signal_style
,pmax.long_signal.signal_key
,pmax.long_signal.signal_color
,pmax.long_signal.signal_marker
,pmax.long_signal.signal_show
,pmax.long_signal.signal_size
,pmax.long_signal.signal_label)
```

## 7. 改信号样式


```python

pmax.long_signal.signal_style=SignalStyle("low",Colors.yellow,Markers.circle_x)
pmax.long_signal.signal_key="low"
pmax.long_signal.signal_color=Colors.blue
pmax.long_signal.signal_marker=Markers.circle_cross
pmax.long_signal.signal_show=True
pmax.long_signal.signal_size=16
pmax.long_signal.signal_label=SignalLabel("buy",12)
pmax.btplot("last")
```

## 8. 改标签


```python
pmax.long_signal.set_label("买入",12)
pmax.btplot("last")
```

## 9. overlap 开关


```python
pmax.short.overlap
pmax.short.overlap=False
pmax.btplot("last")
```

## 10. isplot 开关


```python
pmax.short.isplot
pmax.short.isplot=False
pmax.btplot("last")
```

## 11. follow 开关


```python
pmax.thrend.follow
```

## 小结

- `Line` 是"最小可绘制单位"：改它只影响这一条线；
- 信号是 `Line` 的布尔序列，`set_label()` 控制文字与字号。
