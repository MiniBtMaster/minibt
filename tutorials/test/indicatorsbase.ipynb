# IndicatorsBase 速查：属性 / 方法 / 数据 / 绘图

`BtIndicator` 家族的公共基类能力一览：设置、数据访问、第三方库桥接、画图。

**本次演示**

- `get_indicator_kwargs()` 看指标设置
- `kline` / `source_object`
- `multi_apply()` 批量算
- `ta.pta_sma(...)` 第三方库桥接
- `btplot(height=...)` 画图

不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

## 1. 导入与内置数据


```python
from minibt import *

kline=LocalDatas.test.kline
kline.head()
```

## 2. 造一个 RSI


```python
rsi=kline.close.rsi()
```

## 3. 指标设置


```python
rsi.get_indicator_kwargs() # 指标设置
```

## 4. 所属 K 线 / 类型


```python
rsi.kline is kline,type(rsi)
```

## 5. 属性


```python
# 属性
(rsi.sname
,rsi.ind_name
,rsi.id
,rsi.strategy_id
,rsi.sid
,rsi.plot_id
,rsi.data_id
,rsi.resample_id
,rsi.lines
,rsi.line
,rsi.category
,rsi.iscandles
,rsi.isplot
,rsi.height
,rsi.overlap
,rsi.signallines
,rsi.issignal
,rsi.candle_style
,rsi.ismain
,rsi.isreplay
,rsi.isresample
,rsi.isMDim
,rsi.span_style
,rsi.span_location
,rsi.span_color
,rsi.span_dash
,rsi.span_width
,rsi.isindicator
,rsi.iscustom
,rsi.V
,rsi.length
,rsi.H
,rsi.btindex
,rsi.islivetrading
,rsi.isline
,rsi.new
,rsi.prev
,rsi.sndprev
,rsi.trdprev
,rsi.frthprev
,rsi.strategy_instances
,rsi.strategy_instance)

```

## 6. 方法


```python
# 方法
(rsi.history()
,rsi.upsample()
,rsi.enumerate(kline.close.sma(5),kline.close.sma(10))
,rsi.ifs(kline.close.sma(5)>kline.close.sma(10),1,other=0)
,rsi.ones
,rsi.zeros
,rsi.full()
,rsi.rolling(10).mean()
,rsi.ewm(10).std()
,rsi.expanding(10).median()
,rsi.rolling_apply(lambda x:x*x,10)
,rsi._get_plot_datas(rsi.sname)
,rsi.iloc[:]
,rsi.loc[:]
,rsi.concat([kline.close.ebsw(),kline.close.sma()],axis=0)
,rsi.get_first_valid_index()
,kline.get_max_missing_count(rsi))
```

## 7. multi_apply()


```python
rsi.multi_apply(
        Multiply(PandasTa.ebsw),  # 复杂指标
        [rsi.sma, dict(length=20)], # 简单指标1
        [rsi.sma, dict(length=30)]  # 简单指标2
    )
```

## 8. 数据访问


```python
# 数据
(rsi.plotinfo
,rsi.ind_setting
,rsi.dataset
,rsi.pandas_object
,rsi.kline_object
,rsi.conversion_object
,rsi.custom_object)
```

## 9. 第三方指标集成


```python
# 第三方指标集成
(rsi.ta # 调用的指标返回pandas对象
,rsi.pta # 调用的指标为内置指标
,rsi.talib
,rsi.tulip
,rsi.btind
,rsi.finta
,rsi.tqfunc
,rsi.tqta
,rsi.pairtrading
,rsi.factors)
#,rsi.tradingview)
```

## 10. CoreFunc 调用


```python
(rsi.ta.pta_sma(20,True)
,rsi.pta.sma()
,rsi.talib.SMA()
,rsi.tulip.sma()
,rsi.btind.pmax()
,kline.finta.ADL()
,rsi.tqfunc.sma()
,kline.tqta.AD()
,rsi.pairtrading.bollinger_bands()
,rsi.factors.adaptive_weight_trend()
,kline.tradingview.Adaptive_Trend_Filter())
```

## 11. 画图


```python
# 画图
rsi.btplot("last")
```

## 12. 改高度


```python
kline.height=200
rsi.btplot("last",height=50)
```

## 小结

- 基类把"数据 / 参数 / 样式 / 绘图"四件事都备好了，写指标只需实现 `next()`；
- `ta.*` 是统一的第三方库入口（pta / talib / tulip …）。
