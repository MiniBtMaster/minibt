# IndSeries 全功能速查：单线指标

`IndSeries` 是"一条线"的指标（均线、RSI…）。本篇演示取值、画图、改线型。

**本次演示**

- `kline.close.sma()` 得到 IndSeries
- `btplot("last")` 只画尾部
- `line_style` 就地改线型 / 颜色
- 改完再画一次看效果

**前置条件**：`pip install minibt`（本页只用**内置数据**，
不需要联网 / 账号，也不会下单）。终端若为 GBK，先设 `PYTHONIOENCODING=utf-8`。

**对应文档**：[9.5_indseries.md](../minibt_docs/docs/minibt_data/9.5_indseries.md)


## 1. 导入与内置数据


```python
from minibt import *
kline=LocalDatas.test.kline
kline.head()
```

## 2. 画 SMA


```python
sma=kline.close.sma()
sma,type(sma)
```

## 3. 只画最后一段


```python
sma.btplot('last')
```

## 4. 查看 line_style


```python
(sma.line_style
,sma.line_dash
,sma.line_width
,sma.line_color)
```

## 5. 改成 LineStyle()


```python
sma.line_style=LineStyle()
sma.line_dash=LineDash.dashdot
sma.line_width=3.
sma.line_color=Colors.blue
```

## 6. 再画一次


```python
sma.btplot('last')
```

## 小结

- IndSeries 上几乎所有 pandas Series 方法都能用（见 `pandasseries.ipynb`）；
- `btplot("last")` 适合看最近的形态，`btplot()` 全量。
